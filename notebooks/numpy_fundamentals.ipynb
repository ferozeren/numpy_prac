### 16. Ouput?

```python
import numpy as np

cond = np.array([True, False])
x = np.arange(1, 5).reshape(2, 2)
y = 17
print(np.where(cond, x, y))
```

`````python

cond = np.array([True, False])
x = np.arange(1, 5).reshape(2, 2)
y = 17

print(np.where(cond, x, y))

# np.where() makes an array from elements chosen from x or y depending on condition. 
# The arguments are broadcast to the same shape.

# cond has a shape of (2,) (2 elements)
# x has a shape of (2, 2) (4 elements)
# y has a shape of () (1 scalar element)

# Broadcasting. NumPy automatically stretches smaller arrays to match the size of the largest array.
# To make them match, NumPy stretches cond and y to become (2, 2) matrices.

# cond: When matching a 1D array to a 2D array, NumPy duplicates the row downwards for every row in x.
# The [True, False] line gets repeated for the second row.
# [True False]
# [True False]

# x:
# [1 2]
# [3 4]


# y: The scalar 17 is the easiest. NumPy just copies it to fill every position of a \(2 \times 2\)
# [17 17]
# [17 17]


# np.where(cond, x, y)
# [1 17]
# [3 17]
`````

### 19. Output ?

```python

    lambda x, y: 10 * x + y,
    (3, 3),
)

```

`````python
a = np.fromfunction(
    lambda x, y: 10 * x + y,
    (3, 3),
)

print(a)

# np.fromfunction() makes an array by executing a function over each coordinate.

# The np.fromfunction function creates an array by running your lambda function on the index coordinates of each element.
# The Shape (3, 3): This creates a grid with 3 rows and 3 columns.
# The Inputs x and y: NumPy passes the row indices to x and the column indices to y.

# NumPy generates two internal index matrices for a (3,3) space:
# 1. Get the x and y coordinate grids
x, y = np.indices((3, 3))

print(x, y, sep="\n")

# Row Indices(X):
# [0 0 0]
# [1 1 1]
# [2 2 2]


# Cols Indices(Y):
# [0 1 2]
# [0 1 2]
# [0 2 2]


# -   **Top Row (`x = 0`)**:
#     -   `10*(0) + 0 = 0`
#     -   `10*(0) + 1 = 1`
#     -   `10*(0) + 2 = 2` 
# -   **Middle Row (`x = 1`)**:
#     -   `10*(1) + 0 = 10`
#     -   `10*(1) + 1 = 11`
#     -   `10*(1) + 2 = 12` 
# -   **Bottom Row (`x = 2`)**:
#     -   `10*(2) + 0 = 20`
#     -   `10*(2) + 1 = 21`
#     -   `10*(2) + 2 = 22`
`````

### 22. Output?


`````python
a = np.array(
    [[0, 1],
     [10, 11]]
)

for x in a.flat:
    print(x)

# An array's flat attribute is an np.flatiter instance, which is similar to an iterator over the array's elements.

# Extra

# Use .flat in a for loop to save memory. It does not allocate memory for a new 1D array.

# You can overwrite specific multi-dimensional elements using a single flat index.
arr = np.array([[1,2], [3, 4]])
a.flat[2] = 99
print(arr)

# You can replace multiple items at once using 1D slice logic on a 2D or 3D array.
arr.flat[1:3] = [0, 0]
print(arr)

`````

### 24. Output?

```python
import numpy as np

a = np.arange(8).reshape(2, 4)
b = a.T
print(a.shape, b.shape)
b[0, 0] = 7
print(a)

```

`````python
a = np.arange(8).reshape(2, 4)


# a.T returns a new transposed array, which views the same data as a.
# Rule of thumb: views are returned when possible.


# Creates a transposed view b. Its shape becomes \(4 \times 2\). It points to the exact same flat line in memory, but reads it vertically.
b = a.T

print(a.shape, b.shape)

# You target the top-left element of b. That slot shares the physical memory location with the top-left element of a (which was 0).
# The value in that shared memory cell changes to 7.
b[0, 0] = 7

print(a)
`````

### 25. Output?
```python
import numpy as np

a = np.arange(8).reshape(2, 4)
print(a.ravel())
print(a)
```

`````python
a = np.arange(8).reshape(2, 4)
print(a.ravel())
print(a)

# The ravel() method returns a new flattened array. Not in place!

# Extra

# ravel(): Returns a view by default. It is highly execution-speed efficient and memory-friendly because it skips allocating  new space.

# flatten(): Always forces a deep copy. 
# Changes to your flattened output will never impact your original array data, but it consumes more system memory for large structures

arr = np.array([[1, 2, 3], [4, 5, 6]])

# Both methods yield the exact same result:
print(arr.ravel())  # [1 2 3 4 5 6]

# Shared Reference: If you change an element in the flattened view, the original array mutates:

flat = arr.ravel()
flat[0] = 99
print(arr)  # [[99,  2,  3], [ 4,  5,  6]]
`````

### 27. Output?

```python
import numpy as np

    [[1, 2],
     [3, 4]]

    [[5, 6],
     [7, 8]]
)

```


`````python
a = np.array(
    [[1, 2],
     [3, 4]]
)

b = np.array(
    [[5, 6],
     [7, 8]]
)

print(np.vstack((a, b)))
print(np.hstack((a, b)))

# np.vstack() stacks along arrays' first axes and np.hstack() stacks along arrays' second axes.
# Note the tuples!


# Extra
print()
# np.concatenate() joins a sequence of arrays along an existing axis. 
# You must pass the input arrays as a single container (like a tuple or list), 
# and they must have identical shapes except along the axis you are combining. Note that starting with NumPy 2.0, 
# you can also use np.concat as a convenient shorthand


# Vertical Stacking (axis=0)Combines rows. The arrays must have the same number of columns.
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
result = np.concatenate((a, b), axis=0)
print(result)
# Output:
# [[1 2]
#  [3 4]
#  [5 6]]

# Horizontal Stacking (axis=1)Combines columns. The arrays must have the same number of rows.pythona = np.array([[1, 2], [3, 4]])
a = np.array([[1, 2], [3, 4]])
b = np.array([[5], [6]])

result = np.concatenate((a, b), axis=1)
print(result)
# Output:
# [[1 2 5]
#  [3 4 6]]
`````
